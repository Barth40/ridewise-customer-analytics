```python
import pandas as pd
import numpy as np

# Dictionary containing dataset names and date columns
data_files = {
    "sessions": ["session_time"],
    "trips": ["pickup_time", "dropoff_time"],
    "riders": ["signup_date"],
    "drivers": ["signup_date", "last_active"],
    "promotions": ["start_date", "end_date"]
}

# Dictionary to store DataFrames
dfs = {}

# Load all datasets
for name, date_cols in data_files.items():
    dfs[name] = pd.read_csv(
        f"../data/raw/{name}.csv",
        parse_dates=date_cols
    )

# Assign DataFrames to variables
sessions = dfs["sessions"]
trips = dfs["trips"]
riders = dfs["riders"]
drivers = dfs["drivers"]
promotions = dfs["promotions"]
```

Profiling drivers Dataset

```python
# Create a copy of the original drivers dataset
# This prevents changes from affecting the raw data
drivers_clean = drivers.copy()

# Display the number of rows and columns
print("Original shape:", drivers_clean.shape)

# Display the data types of each column
print("\nData Types:")
print(drivers_clean.dtypes)

# Check for duplicate rows
print("\nDuplicate rows:", drivers_clean.duplicated().sum())

# Check for duplicate Driver IDs
print("Duplicate Driver IDs:",
      drivers_clean["driver_id"].duplicated().sum())

# Check for missing values in each column
print("\nMissing values:")
print(drivers_clean.isnull().sum())

# Preview the first five records
display(drivers_clean.head())
```

```python
# Display descriptive statistics for both numeric
# and categorical columns
display(drivers_clean.describe(include="all"))
```

```python

# Check Date Ranges

# Display the earliest and latest driver signup dates
print("Driver Signup Date Range")
print("-" * 35)
print(f"Earliest Signup Date : {drivers_clean['signup_date'].min()}")
print(f"Latest Signup Date   : {drivers_clean['signup_date'].max()}")

# Display the earliest and latest driver activity dates
print("\nDriver Last Active Date Range")
print("-" * 35)
print(f"First Activity Date  : {drivers_clean['last_active'].min()}")
print(f"Latest Activity Date : {drivers_clean['last_active'].max()}")
```

```python
# Count the number of drivers in each vehicle category
print(drivers_clean["vehicle_type"].value_counts())
```

```python
# Count the number of drivers in each city
print(drivers_clean["city"].value_counts())
```

```python
# Display summary statistics for driver ratings
print(drivers_clean["rating"].describe())
```

```python
# Display summary statistics for acceptance rate
print(drivers_clean["acceptance_rate"].describe())
```

```python
# Ratings should normally fall between 1 and 5

invalid_rating = drivers_clean[
    (drivers_clean["rating"] < 1) |
    (drivers_clean["rating"] > 5)
]

print("Invalid ratings:", len(invalid_rating))
```

```python
# Acceptance rate should be between 0 and 1

invalid_acceptance = drivers_clean[
    (drivers_clean["acceptance_rate"] < 0) |
    (drivers_clean["acceptance_rate"] > 1)
]

print("Invalid acceptance rates:", len(invalid_acceptance))
```

```python
# Calculate the Interquartile Range (IQR)

Q1 = drivers_clean["rating"].quantile(0.25)
Q3 = drivers_clean["rating"].quantile(0.75)

IQR = Q3 - Q1

# Calculate lower and upper limits

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify outliers

outliers = drivers_clean[
    (drivers_clean["rating"] < lower) |
    (drivers_clean["rating"] > upper)
]

print("Rating Outliers:", len(outliers))
```

```python
# Display a summary of the dataset after profiling

print("Original shape:", drivers_clean.shape)
print("Cleaned shape:", drivers_clean.shape)

print("\nDuplicate rows:", drivers_clean.duplicated().sum())

print("Duplicate Driver IDs:",
      drivers_clean["driver_id"].duplicated().sum())

print("\nMissing values:")
print(drivers_clean.isnull().sum())
```

```python

# Numerical Variables

sessions.describe().T
```

```python
# Categorical Variables


exclude_cols = ["session_id", "rider_id"]

for col in sessions.select_dtypes(include=["object", "bool"]).columns:

    if col in exclude_cols:
        continue

    print(f"\nColumn: {col}")
    print("-"*50)

    summary = (
        sessions[col]
        .value_counts(dropna=False)
        .rename("Count")
        .to_frame()
    )

    summary["Percentage (%)"] = (
        summary["Count"] / len(sessions) * 100
    ).round(2)

    print(summary)
```

```python

# Session Date Range

print("Earliest Session:")
print(sessions["session_time"].min())

print()

print("Latest Session:")
print(sessions["session_time"].max())
```

```python
# Numerical Validation

print(
    "Time on App Valid:",
    (sessions["time_on_app"] >= 0).all()
)

print(
    "Pages Visited Valid:",
    sessions["pages_visited"].between(1,5).all()
)

print(
    "Converted Valid:",
    sessions["converted"].isin([0,1]).all()
)
```

```python
# Check Extreme Values

print("Maximum Time on App:")
print(sessions["time_on_app"].max())

print()

print("Minimum Time on App:")
print(sessions["time_on_app"].min())
```

```python

# Conversion Distribution


conversion = pd.DataFrame({
    "Count": sessions["converted"].value_counts(),
    "Percentage (%)":
        (
            sessions["converted"]
            .value_counts(normalize=True)*100
        ).round(2)
})

conversion
```

```python
# Categorical Variable Distribution

exclude_cols = ["promo_id"]

for col in promotions.select_dtypes(include=["object"]).columns:

    if col in exclude_cols:
        continue

    print(f"\nColumn: {col}")
    print("-"*50)

    summary = pd.DataFrame({
        "Count": promotions[col].value_counts(dropna=False),
        "Percentage (%)":
            (
                promotions[col]
                .value_counts(normalize=True)*100
            ).round(2)
    })

    display(summary)
```
